## **Basis of Numpy Library**

Numpy is the core library for scientific computing in Python. It provides a high-performance multidimensional array object and tools for working with these **arrays**.

To use Numpy, we first need to import the `numpy` package:

```python
# import library
import numpy as np
```

### **Arrays**

- A numpy array is a grid of values, all of the same type, and is indexed by a tuple of nonnegative integers.

- The number of dimensions is the rank of the array;

- The shape of an array is a tuple of integers giving the size of the array along each dimension.

**Exercise:**
- What is an **array**?
- Give an example that can be represented by an array.

**Answer:**
- An **array** is a collection of values of the same data type arranged in one or more dimensions.
- Example: the grayscale pixel values of an image can be represented by a 2D array, where each row and column corresponds to a pixel position.


```python
# create 1d array
a = np.array([1, 2, 3, 4])
print(a)
```

```python
# print shape
print(a.shape)
```

```python
# create 2d array
b = np.array([[1, 2, 3],
              [4, 5, 6]])
print(b)
```

```python
# print shape
print(b.shape)
```

```python
# create 3d array
c = np.array([[[1, 2],
               [3, 4]],
              [[5, 6],
               [7, 8]]])
print(c)
```

```python
# print shape
print(c.shape)
```

### **Array Indexing**

```python
# print the second element in your 1d array
print(a[1])
```

```python
# print the first row second element in your 2d array
print(b[0, 1])
```

```python
# print the first rank first row second element in your 3d array
print(c[0, 0, 1])
```

### **Array Operations**

```python
# create 2 2d-array
x = np.array([[1, 2],
              [3, 4]])
y = np.array([[5, 6],
              [7, 8]])

# Elementwise sum; both produce an array
print(x + y)
print(np.add(x, y))
```

```python
# Elementwise difference; both produce an array
print(x - y)
print(np.subtract(x, y))
```

```python
# Elementwise division; both produce an array
print(x / y)
print(np.divide(x, y))
```

```python
# Elementwise product; both produce the array
print(x * y)
print(np.multiply(x, y))
```

```python
# Dot product or matrix multiplication
print(x.dot(y))
print(x @ y)
```
